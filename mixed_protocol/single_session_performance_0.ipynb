# Single-session performance — the random timeout / banishment task

Log-only performance analysis for ONE session, when all you have is `log.json`. This task puts
**one punishment icon** on the board that is **randomly a banishment or a timeout** (its `options`
field lists the pair `[['banish','fountains'],['timeout','target']]`, and the two have DIFFERENT
textures, so the mouse can see which it is), alongside the reward(s).

What this notebook makes (all from the log):

1. a **clean per-collection dataframe** (like the `results/` tables, but for a single session);
2. the **stochastic p-value over all trials** — is he collecting rewards above the world's chance ratio;
3. the **stochastic p-value for the timeout icon only**;
4. **how many times each effect was collected** (reward / timeout / banishment / escape);
5. the **distance and time between collections**;
6. the **path occupancy around each collected effect** (icon-centred);
7. the **heat map centred on a collected icon type**;
8. the **heading error toward each icon, per trial**, in the trial-report style.

Reuses `common/perf_from_log.py` (stochastic p, world opportunity), `common/geom.py`,
`common/heading.py`, `common/viewport.py`. Nothing needs the video.

```python
import sys, json
from pathlib import Path
import numpy as np, pandas as pd
import matplotlib.pyplot as plt

# locate common/ (works from results/ in the repo OR from a session folder like flow_test/mouse2/)
_HERE = Path.cwd()
def _find_common():
    for base in (_HERE, *_HERE.parents):
        for rel in ('common', 'session_pipeline/common'):
            if (base / rel / 'perf_from_log.py').exists():
                return base / rel
    return None
_COMMON = _find_common()
assert _COMMON is not None, f'cannot locate common/ from {_HERE}'
sys.path.insert(0, str(_COMMON))
import perf_from_log as pfl
import geom, heading, viewport as vp
print('common ->', _COMMON)

try:                       # keep figures in memory so save_report() below can collect them all
    get_ipython().run_line_magic('config', 'InlineBackend.close_figures = False')
except Exception:
    pass
```

```python
# ── CONFIG: point at ONE session folder that contains log.json ─────────────────
SESSION_DIR = Path('.')            # <-- the folder with log.json (mouse2 copy: leave as '.')
VIEW_SCALE  = getattr(pfl, 'DEFAULT_VIEW_SCALE', 0.35)   # world zoom (not in the log; 0.35 default)

SESSION_DIR = SESSION_DIR.resolve()
LOGP = SESSION_DIR / 'log.json'
assert LOGP.exists(), f'no log.json in {SESSION_DIR}'
L = json.load(open(LOGP))
W = float(L['worlds'][0].get('width', 2400)); H = float(L['worlds'][0].get('height', 2400))
sess = {'view_scale': VIEW_SCALE, 'world_width': W, 'world_height': H}

# effect valence (effect-level, from perf_from_log)
POS, NEG, NEU = pfl.POSITIVE_EFFECTS, pfl.NEGATIVE_EFFECTS, pfl.NEUTRAL_EFFECTS
def valence(e): return 'positive' if e in POS else 'negative' if e in NEG else 'neutral' if e in NEU else 'other'

t_xy, ax_, ay_ = geom.load_coords(L)           # avatar trajectory (t_ms, x, y)
t_th, theta = geom.load_theta(L)               # avatar heading (t_ms, rad)
theta_t = np.column_stack([t_th, theta]); coords = np.column_stack([t_xy, ax_, ay_])
ed = L.get('experiment_data', {})
print('mouse   :', ed.get('ID'), '| datetime:', ed.get('datetime'))
print('world   :', f'{W:.0f}x{H:.0f}', '| view_scale', VIEW_SCALE, '| collections', len(L['collected']))
import collections as _c
print('effects :', dict(_c.Counter(c['effect'] for c in L['collected'])))
```

## 1 — the clean per-collection dataframe

One row per collection (`trial = one collected icon`), with the disjoint window `(previous collection,
this collection]`: the collected effect + valence, its position, the reward multiplier, whether the
punishment was randomly assigned (`random_punish` = the `options` field was set), and the per-window
geometry (path efficiency / time in corner / mean speed / heading alignment toward the collected icon),
using the same `common/geom` functions the `results/` tables use.

```python
coll = L['collected']
rows = []
prev_t = 0.0
for i, c in enumerate(coll):
    tc = float(c['time']); s_ms, e_ms = prev_t, tc
    tt, xx, yy = geom.slice_track(t_xy, ax_, ay_, s_ms, e_ms)
    msp, _ = geom.speed_stats(tt, xx, yy)
    _, align = geom.heading_to_target(t_th, theta, t_xy, ax_, ay_, (c['x'], c['y']), s_ms, e_ms)
    dist_prev = float(np.hypot(c['x'] - coll[i-1]['x'], c['y'] - coll[i-1]['y'])) if i > 0 else np.nan
    rows.append(dict(
        idx=i, effect=c['effect'], valence=valence(c['effect']), texture=c.get('texture'),
        x=float(c['x']), y=float(c['y']), loc=c.get('loc'), time_ms=tc,
        dt_prev_ms=tc - prev_t, dist_prev=dist_prev,
        multiplier=c.get('multiplier'), duration=c.get('duration'),
        random_punish=c.get('options') is not None,
        path_efficiency=geom.path_efficiency(xx, yy),
        time_in_corner=geom.time_in_corner(xx, yy, W, H, t=tt),
        mean_speed=msp, heading_align=align, start_ms=s_ms, end_ms=e_ms))
    prev_t = tc
df = pd.DataFrame(rows)
OUTP = SESSION_DIR / 'clean_trials_log.pkl'
df.to_pickle(OUTP); print('saved', OUTP, '| shape', df.shape)
df.head(12)
```

## 2 — stochastic p-values

The **world opportunity** baseline: the board offers `active_benefits` good vs `active_detriments` bad
icons, so with NO discrimination the expected hit rate is `benefits/(benefits+detriments)`. The
**stochastic p** = P(≥ the observed number of rewards, if each choice were good with that probability)
— a small p means he collects rewards **above** chance (he discriminates); a large p means at/below
chance. `(1)` pools both punishments; `(2)` counts only the **timeout** icon as the bad option
(reward-vs-timeout), and `(3)` only banishment, so you can see whether he avoids one more than the other.

```python
ben, det, chance = pfl.world_opportunity(L)
n_pos = int((df.valence == 'positive').sum())
n_timeout = int((df.effect == 'timeout').sum()); n_banish = int((df.effect == 'banish').sum())
n_neg = int((df.valence == 'negative').sum())

def stoch(good, bad, label):
    x = good + bad
    p = pfl.prob_at_least(good, x, chance)
    print(f'  {label:28s} reward {good:2d} vs bad {bad:2d}  (rate {good/x:.3f}, chance {chance:.3f})  '
          f'stochastic p = {p:.4f}')
    return p

print(f'world: {ben} benefit / {det} detriment icons  ->  chance ratio = {chance:.3f}\n')
p_all     = stoch(n_pos, n_neg,     '(1) ALL punishments')
p_timeout = stoch(n_pos, n_timeout, '(2) TIMEOUT only')
p_banish  = stoch(n_pos, n_banish,  '(3) BANISHMENT only')
print('\n(escape/unbanish collections are NEUTRAL and excluded from every rate)')

# cross-check with the full log-only scorecard (note: it classifies this board as banish_multiplier)
try:
    R = pfl.score_log(str(LOGP), view_scale=VIEW_SCALE)
    print('\nscore_log cross-check:',
          {k: (round(R[k], 3) if isinstance(R.get(k), float) else R.get(k))
           for k in ('task', 'observed', 'chance_game_design', 'D_game_design', 'p_game_design') if k in R})
except Exception as e:
    print('\n(score_log skipped:', e, ')')
```

## 4d — decision-making test: does he prefer / exploit the multiplier?

The standard 2-choice sequential test. Every collection is a choice between **good** (reward) and
**bad** (a negative — banishment *or* timeout pooled); escapes are neutral and excluded. So the
**overall reward rate** (the dashed line in a & b) = rewards ÷ all good/bad choices = **35 ÷ 51 = 0.69**.

- **(a) Win-stay** — after collecting a reward, how often is his *next* collection also a reward, vs
  after hitting a negative? "After a reward" clearly higher ⇒ he perseveres on reward (win-stay), the
  signature of exploiting a streak reward. (Fisher exact test on the 2×2.)
- **(b) Escalation** — does `P(next = reward)` climb the longer his current reward run is? (Wilson 95 %
  intervals; read the bands, not the dots — n shrinks fast at long runs.)
- **(c) Earnings, in reward-drops.** Three bars, each a NUMBER OF DROPS the pump delivered:
    - **no multiplier** — if every reward paid a flat 1 drop → `35` (he has 35 rewards).
    - **his actual order** — what his real sequence paid, each reward worth its current multiplier → `96`.
    - **one big streak (ceiling)** — if all 35 rewards had arrived in one unbroken run. The multiplier
      climbs 1,2,3,4 and **caps at 4**, so `1+2+3+4×32 = 134` — the most the multiplier could ever pay
      for 35 rewards.
  **Bonus captured = (his − no-multiplier) / (ceiling − no-multiplier) = (96−35)/(134−35) = 62 %** — how
  much of the *available* multiplier bonus he actually banked (0 % = never benefited, 100 % = maximal).

```python
import streaks
from scipy.stats import binomtest, fisher_exact
seq = df[df.valence.isin(['positive', 'negative'])].valence.eq('positive').values   # True = reward
st = streaks.analyse(seq, mult_cap=4)
base = float(seq.mean())
cur, nxt = seq[:-1], seq[1:]
p_after_good = float(nxt[cur].mean());  n_good = int(cur.sum())
p_after_bad  = float(nxt[~cur].mean()); n_bad  = int((~cur).sum())
a, b = int((cur & nxt).sum()), int((cur & ~nxt).sum())
c_, d_ = int((~cur & nxt).sum()), int((~cur & ~nxt).sum())
_, p_fish = fisher_exact([[a, b], [c_, d_]])

def drops(s, cap=4):
    tot, m = 0, 1
    for x in s:
        if x: tot += m; m = min(m + 1, cap)
        else: m = 1
    return tot
n_pos = int(seq.sum())
flat, best, earned = n_pos, drops(np.ones(n_pos, bool)), st['drops_earned']
captured = (earned - flat) / (best - flat) if best > flat else np.nan

fig, ax = plt.subplots(1, 3, figsize=(15, 4.3))
ax[0].bar([0, 1], [p_after_good, p_after_bad], color=[colr['single_reward'], colr['banish']])
ax[0].axhline(base, ls='--', color='k', label=f'overall reward rate {base:.2f}')
for x, v in zip([0, 1], [p_after_good, p_after_bad]): ax[0].text(x, v + 0.02, f'{v:.2f}', ha='center', fontsize=11)
ax[0].set_xticks([0, 1]); ax[0].set_xticklabels([f'after a REWARD\n(n={n_good})', f'after a NEGATIVE\n(n={n_bad})'])
ax[0].set_ylim(0, 1.08); ax[0].set_ylabel('P(next collection = reward)')
ax[0].set_title(f'(a) win-stay  (Fisher p={p_fish:.3f})'); ax[0].legend(fontsize=8)
ks = sorted(st['cond']); pk = [st['cond'][k]['p'] for k in ks]
lo = [st['cond'][k]['p'] - st['cond'][k]['lo'] for k in ks]; hi = [st['cond'][k]['hi'] - st['cond'][k]['p'] for k in ks]
ax[1].errorbar(ks, pk, yerr=[lo, hi], marker='o', capsize=3, color=colr['single_reward'])
for k in ks: ax[1].annotate(f"n={st['cond'][k]['n']}", (k, st['cond'][k]['p']), fontsize=7, textcoords='offset points', xytext=(0, 8), ha='center')
ax[1].axhline(base, ls='--', color='k'); ax[1].set_ylim(0, 1.1)
ax[1].set_xlabel('rewards collected in a row so far'); ax[1].set_ylabel('P(next = reward)'); ax[1].set_title('(b) escalation with the streak')
ax[2].bar([0, 1, 2], [flat, earned, best], color=['0.75', colr['single_reward'], '0.4'])
ax[2].set_xticks([0, 1, 2])
ax[2].set_xticklabels([f'no multiplier\n(1 each)\n{flat}', f'his actual\norder\n{earned}', f'one big streak\n(ceiling)\n{best}'])
ax[2].set_ylabel('reward drops earned'); ax[2].set_title(f'(c) multiplier bonus captured = {captured*100:.0f}%')
plt.suptitle(f"{ed.get('ID')}  decision-making: does he prefer / exploit the multiplier?"); plt.tight_layout(); plt.show()

print('--- in plain words ---')
print(f'Of his {len(seq)} good/bad choices, {n_pos} were rewards ({base:.0%} overall).')
print(f'WIN-STAY: right after a reward he goes to another reward {p_after_good:.0%} of the time,')
print(f'          but only {p_after_bad:.0%} right after hitting a negative (Fisher p={p_fish:.3f})')
print(f'          -> {"he keeps chasing reward once on a roll." if p_fish < 0.05 else "no clear win-stay."}')
print(f'EARNINGS: he banked {earned} reward drops (a "drop" = one squirt of the reward pump).')
print(f'          Those same {n_pos} rewards would pay {flat} with NO multiplier, or {best} if he')
print(f'          collected all {n_pos} back-to-back -> he captured {captured:.0%} of the multiplier bonus.')
print(f'CLUSTERING: his rewards come in longer runs than a random order would give')
print(f'          ({st["runs"]} same-outcome blocks vs ~{st["runs_exp"]:.0f} expected by chance, p={st["runs_p"]:.3f})')
print(f'          -> he collects rewards in bunches rather than evenly mixing in the negatives.')
```

## 5 — path occupancy around each collected effect (icon-centred)

**How it is computed.** For every collection of a given effect, take the avatar `(x, y)` samples logged
within **±`window_s` (default 3 s)** of that collection's timestamp, and subtract the icon's world
position, so the icon sits at the origin `(0, 0)`. Pool these **icon-centred offsets** across *all*
collections of that type and bin them into a 2-D histogram: each cell counts how many avatar **position samples** landed there. The game logs the avatar's
position at a roughly steady rate **while it moves**, so that count ≈ the **time he spent** at that
offset from the icon (bright = he passed through or lingered there often). Caveat: time spent fully
**parked/still is under-counted**, because positions are logged mainly while moving. The star at the origin is the
icon; the bright ring at ~200 world units is the collection radius (he has to reach the icon to collect
it). Comparing the maps tells you whether he approaches a hazard from all sides (like a reward he does
not avoid) or from a restricted direction.

```python
def offsets_around(effect, window_s=3.0):
    ox, oy = [], []
    for _, r in df[df.effect == effect].iterrows():
        m = (t_xy >= r.time_ms - window_s*1000) & (t_xy <= r.time_ms + window_s*1000)
        ox.append(ax_[m] - r.x); oy.append(ay_[m] - r.y)
    return (np.concatenate(ox), np.concatenate(oy)) if ox else (np.array([]), np.array([]))

def occupancy(effect, half=900, bins=45, window_s=3.0, ax=None, title=None):
    ox, oy = offsets_around(effect, window_s)
    ax = ax or plt.gca(); im = None
    if ox.size:
        Hh, xe, ye = np.histogram2d(ox, oy, bins=bins, range=[[-half, half], [-half, half]])
        im = ax.imshow(Hh.T, origin='lower', extent=[-half, half, -half, half], cmap='magma', aspect='equal')
    ax.plot(0, 0, marker='*', ms=16, color='cyan', mec='k')       # the collected icon at centre
    ax.set_title(title or effect); ax.set_xlabel('x - icon (wu)'); ax.set_ylabel('y - icon (wu)')
    return im

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for a, (e, lbl) in zip(axes, [('single_reward', 'reward'), ('banish', 'banishment'), ('timeout', 'timeout')]):
    im = occupancy(e, ax=a, title=f'{lbl}  (n={int((df.effect==e).sum())})')
    if im is not None:
        fig.colorbar(im, ax=a, fraction=0.046, pad=0.04, label='position samples ≈ time spent')
plt.suptitle('path occupancy around the collected icon (±3 s, icon at centre)'); plt.tight_layout(); plt.show()
```

## 6 — heat map centred on one collected icon type

The same computation, zoomed to one type with a colour bar — pick `ICON` and the window. The star is
the icon; brighter = the avatar spent more time there (occupancy).

```python
ICON = 'single_reward'      # <-- 'single_reward' | 'banish' | 'timeout' | 'unbanish'
WINDOW_S, HALF = 3.0, 900
ox, oy = offsets_around(ICON, WINDOW_S)
fig, ax = plt.subplots(figsize=(6.4, 5.6))
if ox.size:
    Hh, xe, ye = np.histogram2d(ox, oy, bins=45, range=[[-HALF, HALF], [-HALF, HALF]])
    im = ax.imshow(Hh.T, origin='lower', extent=[-HALF, HALF, -HALF, HALF], cmap='magma', aspect='equal')
    fig.colorbar(im, ax=ax, label='position samples ≈ time spent')
ax.plot(0, 0, marker='*', ms=18, color='cyan', mec='k', label=f'{ICON} (collected)')
ax.set_title(f"{ed.get('ID')}  occupancy centred on {ICON}  (±{WINDOW_S}s, n={int((df.effect==ICON).sum())})")
ax.set_xlabel('x - icon (world units)'); ax.set_ylabel('y - icon (world units)'); ax.legend(loc='upper right')
plt.tight_layout(); plt.show()
```

## 7 — heading error toward the icon he collected (per type, over time)

**Heading error = the angle between his heading arrow (which is his direction of travel — verified) and
the straight line to the icon.** 0° = facing it, 90° = sideways, 180° = away. Here it is toward **the icon
he actually collected that trial**, grouped by what he collected, over the last 8 s before the collection
(`t = 0`); faint = each trial, bold = mean.

⚠️ It does **not** reach exactly 0° at `t = 0`, and that is expected, not a bug: the icon is collected as
soon as he comes within the **collection radius (~240 wu)** — he never stands on its centre — and while
he is still moving, with the heading arrow naturally jittering ±~30°. So a dip to **~20–35°** means "aimed
straight at it". (Earlier this panel pooled *all* on-screen icons of a type, incl. ones he was not
heading to, which averaged toward ~90°; this version follows only his actual target.)

```python
spawns = sorted(L.get('spawns', []), key=lambda s: s['time'])
def board_at(tc):
    cur = []
    for sp in spawns:
        if sp['time'] <= tc: cur = sp.get('current') or cur
        else: break
    return [ic for ic in cur if ic.get('effect') in (POS | NEG)]

COLL_R = float(np.median([np.hypot(r.x - np.interp(r.end_ms, t_xy, ax_), r.y - np.interp(r.end_ms, t_xy, ay_))
                          for _, r in df.iterrows()]))     # distance at collection (the collection radius)
TYPES = [('single_reward', 'reward'), ('banish', 'banishment'), ('timeout', 'timeout')]
GRID = np.linspace(-8, 0, 80)

def collected_curves_time(effect):
    curves = []
    for i in range(len(df)):
        r = df.iloc[i]
        if r.effect != effect: continue
        e_ms = r.end_ms; s_ms = max(r.start_ms, e_ms - 8000)
        tt, err = heading.heading_error(theta_t, coords, (r.x, r.y), s_ms, e_ms)
        if tt.size: curves.append(((tt - e_ms)/1000, np.degrees(np.abs(err))))
    return curves

fig, axes = plt.subplots(1, 3, figsize=(15, 4.4), sharey=True)
for a, (eff, lbl) in zip(axes, TYPES):
    curves = collected_curves_time(eff); grid_vals = []
    for trel, edeg in curves:
        a.plot(trel, edeg, color=colr[eff], alpha=0.18, lw=.8)
        grid_vals.append(np.interp(GRID, trel, edeg, left=np.nan, right=np.nan))
    if grid_vals:
        M = np.vstack(grid_vals); ok = np.isfinite(M).sum(0) > 0
        mean = np.full(M.shape[1], np.nan); mean[ok] = np.nanmean(M[:, ok], axis=0)
        a.plot(GRID, mean, color=colr[eff], lw=3, label='mean')
    a.axhline(90, ls=':', color='0.6', lw=1); a.set_ylim(0, 180)
    a.set_xlabel('time to collection (s)'); a.legend(fontsize=8)
    a.set_title(f'{lbl}   (n={len(curves)} collections)')
axes[0].set_ylabel('heading error (degrees)\n0 = facing the icon · 90 = sideways · 180 = away')
plt.suptitle(f"{ed.get('ID')}  heading error toward the COLLECTED icon, by type  (collection radius ~{COLL_R:.0f} wu)")
plt.tight_layout(); plt.show()

def heading_page(idx, ax=None):        # per-trial helper (degrees)
    r = df.iloc[int(idx)]; e_ms = r.end_ms; s_ms = max(r.start_ms, e_ms - 12000)
    ax = ax or plt.gca()
    tt, err = heading.heading_error(theta_t, coords, (r.x, r.y), s_ms, e_ms)
    if tt.size: ax.plot((tt - e_ms)/1000, np.degrees(np.abs(err)), lw=2.5, color=colr.get(r.effect, 'k'),
                        label=f'collected {elabel.get(r.effect, r.effect)}')
    ax.set_ylim(0, 180); ax.axhline(90, ls=':', color='0.6')
    ax.set_xlabel('time to collection (s)'); ax.set_ylabel('heading error (deg)')
    ax.set_title(f'trial {int(idx)}: {r.effect}'); ax.legend(fontsize=7)

print('heading error (deg) toward the collected icon AT the moment of collection, by effect:')
_at = []
for _, r in df.iterrows():
    tt, err = heading.heading_error(theta_t, coords, (r.x, r.y), r.end_ms - 200, r.end_ms)
    _at.append((r.effect, np.degrees(np.abs(err)).mean() if err.size else np.nan))
_atdf = pd.DataFrame(_at, columns=['effect', 'deg'])
print(_atdf.groupby('effect')['deg'].agg(['median', 'mean', 'count']).round(1).to_string())
```

## 8 — heading error vs distance to the collected icon (per type)

The same, against **distance to the icon he collected**. Far → near left-to-right; the approach ends at
the **collection radius (~240 wu, dashed vertical)** — he never gets closer.

**Why range matters (the black curve).** Because the avatar and icon have size, he collects whenever his
closest approach falls inside the collection radius `R` — i.e. whenever his heading error is below
`arcsin(R / distance)`. That **"hittable cone"** (black line) is **narrow far away** (he has to aim) and
**widens to 90° at the collection radius** (from close up, almost any heading grazes the icon). So:
points **below the black line** are trajectories that would hit; a mean that hugs the cone means he only
turns toward the icon as late as the geometry forces, while a mean **well below** the cone far out means
he genuinely aimed early. This is why the error is not ~0° at contact — near the icon it *cannot* be.

```python
from scipy.stats import binned_statistic
DMAX = 2000.0
def collected_samples_dist(effect, window_s=8.0):
    dd, ee = [], []
    for i in range(len(df)):
        r = df.iloc[i]
        if r.effect != effect: continue
        e_ms = r.end_ms; s_ms = max(r.start_ms, e_ms - window_s*1000)
        m = (theta_t[:, 0] >= s_ms) & (theta_t[:, 0] <= e_ms)
        if not m.any(): continue
        tt, th = theta_t[m, 0], theta_t[m, 1]
        axp = np.interp(tt, coords[:, 0], coords[:, 1]); ayp = np.interp(tt, coords[:, 0], coords[:, 2])
        bearing = np.arctan2(r.y - ayp, r.x - axp)
        dd.append(np.hypot(r.x - axp, r.y - ayp)); ee.append(np.degrees(np.abs(np.angle(np.exp(1j*(th - bearing))))))
    return (np.concatenate(dd), np.concatenate(ee)) if dd else (np.array([]), np.array([]))

fig, axes = plt.subplots(1, 3, figsize=(15, 4.4), sharey=True)
for a, (eff, lbl) in zip(axes, TYPES):
    dist, err = collected_samples_dist(eff)
    if dist.size:
        a.scatter(dist, err, s=4, color=colr[eff], alpha=0.06)
        mean, edges, _ = binned_statistic(dist, err, statistic='mean', bins=25, range=(0, DMAX))
        a.plot((edges[:-1]+edges[1:])/2, mean, color=colr[eff], lw=3, label='mean')
    _dg = np.linspace(COLL_R, DMAX, 200); _cone = np.degrees(np.arcsin(np.clip(COLL_R/_dg, 0, 1)))
    a.plot(_dg, _cone, color='k', lw=1.3, alpha=.8, label='hittable cone  arcsin(R/d)')
    a.axvline(COLL_R, ls='--', color='0.4', lw=1.2); a.axhline(90, ls=':', color='0.6', lw=1)
    a.set_ylim(0, 180); a.set_xlim(DMAX, 0)
    a.set_xlabel('distance to collected icon (wu)   [near →]'); a.legend(fontsize=8)
    a.set_title(f'{lbl}   (n={dist.size} samples)')
axes[0].set_ylabel('heading error (degrees)\n0 = facing the icon · 90 = sideways · 180 = away')
plt.suptitle(f"{ed.get('ID')}  heading error vs distance to the COLLECTED icon  (dashed = collection radius ~{COLL_R:.0f} wu)")
plt.tight_layout(); plt.show()
```

```python
cats = [e for e in ('single_reward', 'banish', 'timeout', 'unbanish') if (df.effect == e).any()]
data = [df[df.effect == e]['path_efficiency'].dropna().values for e in cats]
fig, ax = plt.subplots(figsize=(8, 4.6))
bp = ax.boxplot(data, patch_artist=True, showfliers=False, widths=0.6)
for patch, e in zip(bp['boxes'], cats): patch.set_facecolor(colr[e]); patch.set_alpha(.75)
for med in bp['medians']: med.set_color('k')
rng = np.random.default_rng(0)
for i, v in enumerate(data):
    ax.scatter(np.full(len(v), i + 1) + rng.uniform(-.12, .12, len(v)), v, s=18, color='0.2', alpha=.55, zorder=3)
ax.set_xticks(range(1, len(cats) + 1)); ax.set_xticklabels([elabel[e] for e in cats])
ax.set_ylim(0, 1.02); ax.set_ylabel('path efficiency  (1 = straight beeline, 0 = wandering)')
ax.set_title(f"{ed.get('ID')}  approach path efficiency by collected type")
plt.tight_layout(); plt.show()
print(df.groupby('effect')['path_efficiency'].agg(['median', 'mean', 'count']).round(3).to_string())
```

## 10 — occupancy by TIME (milliseconds)

The same maps as section 5, but the colour is the **total time in milliseconds** the avatar spent at each
offset from the icon, not a raw sample count. Every position sample is weighted by the gap to the next
sample (the time it stands for) and those gaps are summed per cell, pooled over all collections of that
type. Unlike the sample-count map, this **credits pauses correctly** — a still moment logs few samples
but represents real time. Colour = milliseconds spent.

```python
def offsets_around_timed(effect, window_s=3.0):
    ox, oy, dt = [], [], []
    for _, r in df[df.effect == effect].iterrows():
        m = (t_xy >= r.time_ms - window_s*1000) & (t_xy <= r.time_ms + window_s*1000)
        if m.sum() < 2: continue
        tt = t_xy[m]
        d = np.diff(tt); d = np.append(d, np.median(d))     # ms each sample stands for (last -> median)
        ox.append(ax_[m] - r.x); oy.append(ay_[m] - r.y); dt.append(d)
    if not ox: return np.array([]), np.array([]), np.array([])
    return np.concatenate(ox), np.concatenate(oy), np.concatenate(dt)

def occupancy_ms(effect, half=900, bins=45, window_s=3.0, ax=None, title=None):
    ox, oy, dt = offsets_around_timed(effect, window_s)
    ax = ax or plt.gca(); im = None
    if ox.size:
        Hh, xe, ye = np.histogram2d(ox, oy, bins=bins, range=[[-half, half], [-half, half]], weights=dt)
        im = ax.imshow(Hh.T, origin='lower', extent=[-half, half, -half, half], cmap='magma', aspect='equal')
    ax.plot(0, 0, marker='*', ms=16, color='cyan', mec='k')
    ax.set_title(title or effect); ax.set_xlabel('x - icon (wu)'); ax.set_ylabel('y - icon (wu)')
    return im

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for a, (e, lbl) in zip(axes, [('single_reward', 'reward'), ('banish', 'banishment'), ('timeout', 'timeout')]):
    im = occupancy_ms(e, ax=a, title=f'{lbl}  (n={int((df.effect==e).sum())})')
    if im is not None:
        fig.colorbar(im, ax=a, fraction=0.046, pad=0.04, label='time spent (ms)')
plt.suptitle('TIME-weighted occupancy around the collected icon (±3 s, ms spent, icon at centre)')
plt.tight_layout(); plt.show()
```

## 11 — stochastic p-value summary

The headline discrimination numbers in one place. Each bar is the reward rate for a good-vs-bad
comparison; the dashed line is the world's chance ratio (2 good : 1 bad → 0.67). The **stochastic p** =
P(collecting at least this many rewards by chance) — **small p ⇒ he collects reward above chance / avoids
that bad type; p near 1 ⇒ at chance.**

1. **reward vs ALL negatives** (banishment + timeout pooled)
2. **reward vs TIMEOUT only**
3. **reward vs BANISHMENT only**

```python
_ben, _det, _chance = pfl.world_opportunity(L)
_npos = int((df.valence == 'positive').sum())
_ntime = int((df.effect == 'timeout').sum()); _nban = int((df.effect == 'banish').sum())
comps = [('reward vs ALL negatives\n(banishment + timeout)', _npos, _ntime + _nban),
         ('reward vs TIMEOUT only', _npos, _ntime),
         ('reward vs BANISHMENT only', _npos, _nban)]
fig, ax = plt.subplots(figsize=(8.5, 4.6)); rows = []
for i, (lbl, g_, b_) in enumerate(comps):
    x = g_ + b_; rate = g_ / x if x else np.nan; p = pfl.prob_at_least(g_, x, _chance)
    ax.bar(i, rate, color=colr['single_reward'], width=0.6)
    ax.text(i, rate + 0.02, f'{rate:.2f}\np={p:.4f}', ha='center', fontsize=9)
    rows.append((lbl.replace(chr(10), ' '), g_, b_, round(rate, 3), round(p, 4)))
ax.axhline(_chance, ls='--', color='k', label=f'world chance {_chance:.2f}')
ax.set_xticks(range(len(comps))); ax.set_xticklabels([c[0] for c in comps], fontsize=8)
ax.set_ylim(0, 1.08); ax.set_ylabel('reward rate  =  reward / (reward + bad)')
ax.set_title(f"{ed.get('ID')}  stochastic p-values: is reward collected above chance?"); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()
print(pd.DataFrame(rows, columns=['comparison', 'n_reward', 'n_bad', 'reward_rate', 'stochastic_p']).to_string(index=False))
```

## — export: save the whole report as a PDF or PNGs

Run this **after `Run All`** (a fresh run, so the figures above are not duplicated). `save_report('pdf')`
writes every figure, in order, into one multi-page PDF; `save_report('png')` writes them as individual
PNG files. Pass `path=` to redirect (a `.pdf` file, or a folder for the PNGs).

```python
_stem = str(ed.get('ID', 'session')).replace(' ', '_')
def save_report(fmt='pdf', path=None, dpi=130):
    '''Save every figure produced above (in creation order) as ONE PDF, or as individual PNGs.'''
    from matplotlib.backends.backend_pdf import PdfPages
    figs = [plt.figure(n) for n in plt.get_fignums()]
    if not figs:
        print('no open figures -- do "Run All" first, then run this cell'); return
    if fmt == 'pdf':
        p = Path(path) if path else SESSION_DIR / f'{_stem}_performance.pdf'
        p.parent.mkdir(parents=True, exist_ok=True)
        with PdfPages(p) as pdf:
            for f in figs: pdf.savefig(f, bbox_inches='tight')
        print(f'saved {len(figs)}-page PDF -> {p}')
    else:
        d = Path(path) if path else SESSION_DIR / 'report_png'
        d.mkdir(parents=True, exist_ok=True)
        for i, f in enumerate(figs, 1):
            f.savefig(d / f'{_stem}_fig{i:02d}.png', dpi=dpi, bbox_inches='tight')
        print(f'saved {len(figs)} PNGs -> {d}')

# save_report('pdf')      # <- uncomment to write the PDF
# save_report('png')      # <- or the PNGs
print('report export ready: call  save_report("pdf")  or  save_report("png")')
```
